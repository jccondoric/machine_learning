# Cargar Dataset

```python
# Cargar el dataset desde Google Drive
ruta = '/content/drive/MyDrive/machine learning/datasets/Salary_Data.csv'

with open(ruta, 'r') as archivo:
    lineas = archivo.readlines()

print("Total de líneas (incluyendo encabezado):", len(lineas))
print("Encabezado:", lineas[0].strip())
print("Primeras 5 filas de datos:")
for linea in lineas[1:6]:
    print(linea.strip())
```

**Convirtiendo texto (string) a float.**

```python
X = []
Y = []

for linea in lineas[1:]:
    linea = linea.strip()
    partes = linea.split(',')

    experiencia = float(partes[0])
    salario = float(partes[1])

    X.append(experiencia)
    Y.append(salario)

n = len(X)
print("\nCantidad de datos:", n)
print("Primeros 5 valores de X:", X[:5])
print("Primeros 5 valores de Y:", Y[:5])
```

# BÚSQUEDA DE b Y w (ajuste iterativo)

```python
# Modelo:  y = b + w*x
# Costo:   J = (1/n) * sum((b + w*x_i - y_i)^2)
# Gradiente respecto a b:  dJ/db = (2/n) * sum(error_i)
# Gradiente respecto a w:  dJ/dw = (2/n) * sum(error_i * x_i)

b, w = 0.0, 0.0
tasa = 0.01
error_ant = 1e18
for iteracion in range(20000):
    error, grad_b, grad_w = 0.0, 0.0, 0.0

    for i in range(n):
        dif = (b + w * X[i]) - Y[i]
        error  += dif * dif
        grad_b += dif
        grad_w += dif * X[i]
    error /= n
    grad_b *= 2 / n
    grad_w *= 2 / n

    b -= tasa * grad_b
    w -= tasa * grad_w

    if abs(error_ant - error) < 1e-6:
        print(f"Convergencia alcanzada en la iteración {iteracion}")
        break
    error_ant = error

print(f"\nMejor b (intercepto): {round(b, 2)}")
print(f"Mejor w (pendiente):   {round(w, 2)}")
print(f"Error final (costo J): {round(error, 2)}\n")
```

# Inferencia con 10 nuevos valores

```python
print("Inferencias sobre 10 nuevos valores de x (años de experiencia):\n")
nuevos_x = [0.5, 2.5, 4.3, 5.5, 6.2, 7.0, 8.5, 9.9, 10.8, 11.5]
for x in nuevos_x:
    prediccion = b + w * x
    print(f"x = {x:5.1f} años -> salario estimado = ${prediccion:,.2f}")
```

```python
print(f"""
EXPLICACIÓN DEL ALGORITMO EMPLEADO Y DE LOS RESULTADOS
========================================================

ALGORITMO EMPLEADO: Descenso por Gradiente (Batch Gradient Descent)

El descenso por gradiente es un algoritmo iterativo de optimización que
permite encontrar los valores de b y w que minimizan el error entre las
predicciones del modelo y los datos reales, sin necesidad de resolver
una ecuación cerrada.

Modelo (hipótesis):
    y = b + w*x

Función de costo (error cuadrático medio), que mide qué tan mal está
prediciendo el modelo con los valores actuales de b y w:
    J(b,w) = (1/n) * sum( (b + w*x_i - y_i)^2 )

Se eleva al cuadrado el error para que los errores positivos y negativos
no se cancelen entre sí, y para penalizar más los errores grandes que
los pequeños.

Gradientes (derivadas parciales de J respecto a cada parámetro), que
indican la dirección en la que el costo crece más rápido:
    dJ/db = (2/n) * sum( b + w*x_i - y_i )
    dJ/dw = (2/n) * sum( (b + w*x_i - y_i) * x_i )

Regla de actualización, aplicada en cada iteración:
    b := b - alpha * dJ/db
    w := w - alpha * dJ/dw

FUNCIONAMIENTO

El algoritmo parte de b=0 y w=0 (una recta plana, sin información) y en
cada iteración:
  1. Recorre las {n} observaciones del dataset y calcula el error de
     predicción de cada una.
  2. Acumula esos errores para obtener el gradiente respecto a b y
     respecto a w (las sumatorias de la fórmula).
  3. Actualiza b y w moviéndolos en dirección CONTRARIA al gradiente,
     ya que el gradiente apunta hacia donde el error AUMENTA, y
     buscamos justamente lo contrario: que el error disminuya.
  4. El tamaño de cada paso está controlado por la tasa de aprendizaje
     alpha ({tasa} en este programa). Un alpha muy grande puede hacer
     que el algoritmo "salte" el mínimo y no converja; uno muy pequeño
     hace que el entrenamiento sea innecesariamente lento.
  5. Este proceso se repite hasta que el error deja de cambiar de forma
     significativa entre una iteración y la siguiente (diferencia menor
     a 1e-6), lo que indica que el algoritmo ya llegó (o está muy cerca)
     del mínimo de la función de costo. A esto se le llama convergencia.

En este programa, la convergencia se alcanzó en la iteración {iteracion}
(de un máximo de 20000 permitidas), lo que confirma que el algoritmo
encontró una solución estable antes de agotar el límite de iteraciones.

INTERPRETACIÓN DE LOS RESULTADOS OBTENIDOS

- b = {round(b, 2)}: es el salario base teórico que el modelo predice
  para una persona con 0 años de experiencia.
- w = {round(w, 2)}: es el incremento promedio de salario por cada año
  adicional de experiencia. Es decir, según este modelo, cada año de
  experiencia aporta aproximadamente ${round(w, 2)} al salario esperado.
- Costo final = {round(error, 2)}: representa el error cuadrático medio
  residual del modelo. Al estar en unidades de dólares al cuadrado, este
  número por sí solo no es muy intuitivo, pero su raíz cuadrada
  (aprox. ${round(error**0.5, 2)}) da una idea aproximada de cuánto se
  desvían, en promedio, las predicciones del modelo respecto a los
  salarios reales.

Las predicciones sobre los 10 valores nuevos de x muestran una relación
lineal creciente y consistente: a mayor años de experiencia, mayor
salario estimado, siguiendo fielmente la fórmula aprendida y = b + w*x.
Estos 10 valores no fueron usados durante el entrenamiento (a diferencia
de las {n} filas del dataset original), por lo que representan un caso
real de inferencia: usar un modelo ya entrenado para predecir sobre
datos nuevos que nunca vio.
""")
```
